boolean operations

```python
arr = np.array([1,2,3,4,5])
b = arr > 2
print(b)
print(arr[b])
```

```python
#odd numbers are replaced with -1, even numbers remain unchanged
arr = np.array([1,2,3,4,5,])
new_arr = np.where(arr % 2 ==0, arr , -1)
print(new_arr)
```

aggregate function

```python
matrix = np.array([[1,2],[3,4]])
print(np.mean(matrix))
print(np.max(matrix))
print(np.sum(matrix))
```
