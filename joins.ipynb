GroupBy

groups data based on one or more columns and perform aggregate functions

```python
data  = {
    "Department":["HR","IT","HR","IT","Sales"],
    "Salary":[30000,50000,350000,55000,40000]}
df = pd.DataFrame(data)
df
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
print(df.groupby("Department")["Salary"].agg(["sum","mean","max","min","count"]))
```

Apply  function

```python
df = pd.DataFrame({
    "Marks":[80,90,95,40,39,4]})
df["Bonus"] = df["Marks"].apply(lambda x: x+5)
print(df)
```

```python
def passm(x):
    if x>=40:
        return 'pass'
    else:
        return 'fail'
```

```python
df["Result"] = df["Marks"].apply(passm)
print(df)
```

```python
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x>=40 else "Fail"
)
print(df)
```

Date and Time


```python
data = {'time': ['2025-01-27 14:45:30', '2025-02-05 0:15:00' , '2025-03-10 20:30:43']}
df =pd.DataFrame(data)
df
```

```python
df.dtypes
```

```python
#convert to datetime
df['time'] = pd.to_datetime(df['time'])
```

```python
df.dtypes
```

```python
#extract year
df['year'] = df['time'].dt.year

#extract month
df['month'] = df['time'].dt.month

#extract day
df['day'] = df['time'].dt.day

#extract hour
df['hour'] = df['time'].dt.hour

#extract minute
df['minute'] = df['time'].dt.minute

#extract second
df['second'] = df['time'].dt.second

#extract day of the week(mndy = 0,sundy =6)
df['Dayofweek'] =df['time'].dt.dayofweek

#extract day of year
df['Dayofyear'] =df['time'].dt.dayofyear

#extract week number of the year:
df['week'] =df['time'].dt.isocalendar().week

#extract quarter 
df['quarter'] = df['time'].dt.quarter

#get the name of the day(eg : monday)
df['weekday'] = df['time'].dt.day_name()
```

```python
df
```

Data Transformation

converts data into a suitable format for analysis

```python
#reanme column
df.rename(columns ={"Age":"Students_age"},inplace= True)
print(df)
```

```python
df.Students_age.dtypes
```

```python
#change datatype
df["Students_age"]=df["Students_age"].astype("int32")
print(df.dtypes)
```

```python
#create new column
df["age_after_5_years"]=df["Students_age"] + 5
print(df)
```

```python
#replace values
df = pd.DataFrame({
    "Gender":["M","F","M"]
})
df["Gender"] = df["Gender"].replace({
    "M" :"Male",
    "F":"Female"
})
print(df)
                            
```

```python
#convert text to uppercase
df = pd.DataFrame({
    "Name":["Anu","arun","meera"]
})
df["Name1"] = df["Name"].str.upper()
print(df)
```

```python
#convert to lower case
df["Name2"] = df["Name"].str.lower()
print(df)
```

```python
#remove spaces
df = pd.DataFrame({
    "Name":["  Anu","  Arun","  Meera"]
})
df["Name"] =df["Name"].str.strip()
print(df)
```
