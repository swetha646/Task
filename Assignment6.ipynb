```python
df.columns=df.columns.str.lower()
print(df.columns)
```

```python
df.columns.str.replace(" ", "_",regex=False)
print(df.columns)
```

```python
df["name"] = df["name"].str.title()
df["department"] = df["department"].str.title()
df["city"] = df["city"].str.title()
df["gender"] = df["gender"].str.title()
df["performance"] = df["performance"].str.title()
df["email"] = df["email"].str.title()

print(df)
```

```python
df["age"] = df["age"].replace({
    "Thirty": 30,
    "27": 27
})
df
```

```python
df["age"] = df["age"].fillna(df["age"].mean())
df
```

```python
df["age"] = df["age"].astype(int)
df
```

```python
df["salary"] = df["salary"].replace(",", "",regex = True)
df
```

```python
df["salary"] = df["salary"].astype(float)
```

```python
df["salary"] = df["salary"].fillna(df["salary"].median())
df
```

```python
df["department"] = df["department"].fillna(df["department"].mode()[0])
df
```

```python
df["city"]=df["city"].fillna("Unknown")
df
```

```python
df["gender"] = df["gender"].fillna("Unknown")
df
```

```python
df["performance"] = df["performance"].fillna("Unknown")
df
```

```python
df["email"] = df["email"].fillna("Unknown")
df
```

```python
df["joindate"] = df["joindate"].str.replace("/", "-", regex=False)
df
```

```python
df["joindate"] = df["joindate"].fillna(df["joindate"].mode()[0])
df
```

```python
df.drop_duplicates()
```

```python
def salary_category(salary):
    if salary < 40000:
        return "Low"
    elif salary <= 60000:
        return "Medium"
    else:
        return "High"

df["salary_category"] = df["salary"].apply(salary_category)

print(df)
```
