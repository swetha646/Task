```python
import numpy as np
```

```python
#covert from 1d to other d (reshape)
a = np.array([1,2,3,4,5,6])
a.shape
```

```python
#1D to 2D
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3) #(1,6)(6,1)(3,2)
print(a)
print(b)
```

```python
#1D to 3D
c = np.array([1,2,3,4,5,6])
d = c.reshape(1,2,3)
print(d)
```

```python
#2D to 1D
a = np.array([[1,2],[3,4],[6,7],[9,8]])#(4,2)
e = a.reshape(8,)
print(e)
```

```python
#2D to 3D
b = np.array([[1,2],[3,4],[5,6],[7,8]])
d = b.reshape(1,4,2)
print(d)
```

```python
#3D to 2D
p = np.array([[[9,9],[5,5],[4,4],[2,2]]])
s = p.reshape(4,2)
print(s)
```

```python
#3D to 1D
q = np.array([[[9,9],[5,5],[1,1],[0,0]]])
l = q.reshape(8,)
print(l)
```

```python
q = np.array([[[9,9],[5,5],[1,1],[0,0]]])
print(q)
```

```python
#covert any D to 1D
q.flatten()
```

```python
q.reshape(-1)
```

```python
#mathematical optns
a = np.array([1,2])
b = np.array([7,8])
c= np.add(a,b)
print(c)
```

```python
c = np.dot(a,b)
print(c)
```

```python
c = np.subtract(a,b)
c
```

```python
c = np.multiply(a,b)
c
```
