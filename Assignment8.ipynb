```python
import  matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\USER\Downloads\d4bdb23b-4858-482e-9d51-d9fa09faeec4.xlsx")
df
```

```python
plt.plot(df["Month"],df["Sales"],marker='x')
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
plt.bar(df["Month"],df["Profit"])
plt.title("Monthly profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

```python
plt.scatter(df["Marketing_Spend"],df["Sales"])
plt.title("Marketing_spend and sales")
plt.grid()
plt.xlabel("Marketing_spend")
plt.ylabel("Sales")
plt.show()
```

```python
plt.hist(df["Sales"],bins=6)
plt.title("Distribution of sales")
plt.xlabel("Sales")
plt.show()
```
