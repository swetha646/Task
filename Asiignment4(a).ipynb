```python
#Sort the dataset by final_exam_score in descending order and display the top 10 students.
print(df.sort_values(by="final_exam_score", ascending=False).head(10))
```

```python
print(df["parental_education"].unique())
```

```python
#Display rows 10 to 20 and columns:gender,parental_education,previous_grade,final_exam_score
print(df.loc[10:20, ["gender", "parental_education", "previous_grade", "final_exam_score"]])
```

```python
#Display only the students whose gender is female, showing only:gender,study_time_hours,previous_grade,final_exam_score
print(df.loc[df["gender"]=="Female",
            ["gender","study_time_hours","previous_grade","final_exam_score"]])
```

```python
#Convert the final_exam_score column into float and verify its datatype.
df["final_exam_score"]=df["final_exam_score"].astype(float)

print(df["final_exam_score"].dtype)
```

```python
#Remove the internet_access column.
df=df.drop("internet_access",axis=1)

print(df)
```

```python
#Sort the dataset by previous_grade in descending order and display only:gender,previous_grade,parental_education
print(df.sort_values(by="previous_grade",ascending=False)
      [["gender","previous_grade","parental_education"]])
```

```python
#Remove the part_time_job column and display only the first 15 rows using iloc
df=df.drop("part_time_job",axis=1)

print(df.iloc[:15])
```

```python
#Convert previous_grade to float. Then display only students whose previous_grade is greater than 80 using loc.
df["previous_grade"]=df["previous_grade"].astype(float)

print(df.loc[df["previous_grade"]>80])
```

```python
#Merge the Student Performance dataset with the following DataFrame using the gender column.
school=pd.DataFrame({
    "gender":["Female","Male"],
    "Scholarship":["Yes","No"]
})

merged_df=pd.merge(df,school,on="gender")

print(merged_df)
```
