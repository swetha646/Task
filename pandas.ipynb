```python
import pandas as pd
```

```python
marks = [7,4,3,2]
s = pd.Series(marks)
print(s)
```

```python
s = pd.Series(marks,index = ["A","B","C","D"])
print(s)
```

```python
a=pd.Series([1,2,3,4,5],dtype=float)
print(a)
```

```python
a=pd.Series([2,3,4,5,5]).astype(float)
print(a)
```

```python
s = pd.Series([5,7,4,98])
print(s[0])
print(s[1])
```

```python
s=pd.Series([29,6,25,])
print(s + 5)
print(s * 2)
```

```python
#common series functions
print(s.sum())
print(s.mean())
print(s.max())
print(s.min())
print(s.count())
```
