```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\USER\Downloads\c4fe3a02-800a-48ae-b5bc-15bbc219f46a.xlsx")
df
```

```python
plt.plot(df["Month"],df["Sales"],marker='x')
plt.title("Month and Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(df["Month"],df["Sales"])
plt.title("Month and Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.pie(df["Profit"],labels=df["Month"],autopct="%1.1f%%")
plt.title("Profit Distribution by Month")
plt.show()
```

```python
plt.scatter(df["Customers"], df["Sales"])
plt.title("Customers vs Sales")
plt.xlabel("Customers")
plt.ylabel("Sales")
plt.show()
```
