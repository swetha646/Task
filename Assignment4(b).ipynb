```python
import pandas as pd
```

```python
def age_category(age):
    if age < 25:
        return "Young"
    elif age <= 35:
        return "Adult"
    else:
        return "Senior"

df["Age_Category"] = df["Age"].apply(age_category)

print(df)
```

```python
def travel_status(travel):
    if travel == "Travel_Rarely":
        return "Regular"
    elif travel == "Travel_Frequently":
        return "Frequent"
    else:
        return "No Travel"

df["Travel_Status"] = df["BusinessTravel"].apply(travel_status)

print(df)
```

```python
def attrition_status(status):
    if status == "Yes":
        return "Left Company"
    else:
        return "Working"

df["Attrition_Status"] = df["Attrition"].apply(attrition_status)

print(df)
```

```python
df["Department"] = df["Department"].apply(lambda x: x.upper())

print(df)
```

```python
df["DailyRate_Bonus"] = df["DailyRate"] * 1.10

print(df)
```

```python
def distance_category(distance):
    if distance <= 5:
        return "Near"
    elif distance <= 15:
        return "Medium"
    else:
        return "Far"

df["Distance_Category"] = df["DistanceFromHome"].apply(distance_category)

print(df)
```

```python
def experience(years):
    if years >= 10:
        return "Experienced"
    else:
        return "Fresher"

df["Experience_Status"] = df["TotalWorkingYears"].apply(experience)

print(df)
```
